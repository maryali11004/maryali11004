```python
import torch
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from torchvision.ops import nms
from ultralytics import YOLO

# Load both models
model_plantdoc = YOLO("best_plantdoc.pt")  # PlantDoc trained model
model_pest24 = YOLO("best_pest24.pt")      # Pest24 trained model

# Define a function to run inference and get predictions
def get_predictions(model, source, conf_threshold=0.5):
    results = model.predict(source=source, conf=conf_threshold)
    predictions = results[0].boxes  # Get bounding boxes
    boxes = predictions.xyxy.cpu()  # Get coordinates in xyxy format directly
    conf = predictions.conf.cpu()   # Get confidence scores
    cls = predictions.cls.cpu()     # Get class labels
    return boxes, conf, cls

# Class names (replace with your actual class names)
plantdoc_classes = ["disease1", "disease2", "disease3"]  # Add your PlantDoc class names
pest24_classes = ["pest1", "pest2", "pest3"]  # Add your Pest24 class names

# Run inference on the image
source_image = "/content/pest24-maryam-1/test/images/0000100_jpg.rf.bd98f6d64caeefc2495a34880f075305.jpg"
boxes_plantdoc, conf_plantdoc, cls_plantdoc = get_predictions(model_plantdoc, source_image)
boxes_pest24, conf_pest24, cls_pest24 = get_predictions(model_pest24, source_image)

# Check the results to ensure both models are producing predictions
print(f"PlantDoc Results: Boxes: {boxes_plantdoc.shape}, Confidences: {conf_plantdoc.shape}, Classes: {cls_plantdoc.shape}")
print(f"Pest24 Results: Boxes: {boxes_pest24.shape}, Confidences: {conf_pest24.shape}, Classes: {cls_pest24.shape}")

# Create model identifier for each detection (0 for plantdoc, 1 for pest24)
model_id_plantdoc = torch.zeros(len(cls_plantdoc))
model_id_pest24 = torch.ones(len(cls_pest24))

# Combine predictions
combined_boxes = torch.cat((boxes_plantdoc, boxes_pest24), dim=0)
combined_conf = torch.cat((conf_plantdoc, conf_pest24), dim=0)
combined_cls = torch.cat((cls_plantdoc, cls_pest24), dim=0)
combined_model_ids = torch.cat((model_id_plantdoc, model_id_pest24), dim=0)

# Load image for visualization
image = cv2.imread(source_image)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
height, width = image_rgb.shape[:2]

# Colors for different models
plantdoc_color = (0, 255, 0)  # Green for PlantDoc
pest24_color = (255, 0, 0)    # Red for Pest24

# Display the combined predictions on the image
for i in range(len(combined_boxes)):
    box = combined_boxes[i].numpy()
    conf = combined_conf[i].item()
    cls_id = int(combined_cls[i].item())
    model_id = int(combined_model_ids[i].item())

    # Select color based on model
    color = plantdoc_color if model_id == 0 else pest24_color

    # Get class name
    if model_id == 0:
        class_name = f"PlantDoc-{cls_id}" if cls_id < len(plantdoc_classes) else f"PlantDoc-{cls_id}"
    else:
        class_name = f"Pest24-{cls_id}" if cls_id < len(pest24_classes) else f"Pest24-{cls_id}"

    # Draw the bounding box and label
    x1, y1, x2, y2 = map(int, box)
    label = f"{class_name}: {conf:.2f}"

    cv2.rectangle(image_rgb, (x1, y1), (x2, y2), color, 2)

    # Create a filled rectangle for text background
    text_size = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.6, 2)[0]
    cv2.rectangle(image_rgb, (x1, y1 - text_size[1] - 10), (x1 + text_size[0], y1), color, -1)

    # Add text with white color
    cv2.putText(image_rgb, label, (x1, y1 - 5), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

# Show the image with bounding boxes
plt.figure(figsize=(12, 12))
plt.imshow(image_rgb)
plt.title(f"Combined Detection Results\nGreen: PlantDoc | Red: Pest24")
plt.axis('off')
plt.tight_layout()
plt.show()

# Print detection counts
print(f"Number of PlantDoc detections: {len(cls_plantdoc)}")
print(f"Number of Pest24 detections: {len(cls_pest24)}")
print(f"Total combined detections: {len(combined_boxes)}")
```

```python
import torch
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import numpy as np
import os

# Function for debugging YOLO model predictions
def debug_yolo_model(model_path, image_path, model_name, conf_threshold=0.2):
    """
    Thoroughly examine and debug a YOLO model's predictions on a given image.

    Args:
        model_path: Path to the YOLO model
        image_path: Path to the test image
        model_name: Name of the model for display purposes
        conf_threshold: Confidence threshold for predictions
    """
    print(f"\n{'='*50}")
    print(f"DEBUGGING {model_name.upper()}")
    print(f"{'='*50}")

    # 1. Check model file existence
    if not os.path.exists(model_path):
        print(f"ERROR: Model file '{model_path}' does not exist!")
        return

    # 2. Check image file existence
    if not os.path.exists(image_path):
        print(f"ERROR: Image file '{image_path}' does not exist!")
        return

    print(f"Loading model from: {model_path}")
    print(f"Testing on image: {image_path}")

    try:
        # 3. Load model
        model = YOLO(model_path)
        print(f"Model loaded successfully: {type(model)}")

        # 4. Get model task and number of classes
        model_task = model.task
        num_classes = model.model.nc if hasattr(model.model, 'nc') else "Unknown"
        print(f"Model task: {model_task}")
        print(f"Number of classes: {num_classes}")

        # 5. Run prediction with verbose output
        print(f"\nRunning prediction with confidence threshold: {conf_threshold}")
        results = model.predict(source=image_path, conf=conf_threshold, verbose=True)

        # 6. Check if any predictions were made
        boxes = results[0].boxes
        num_detections = len(boxes)
        print(f"\nDetections found: {num_detections}")

        if num_detections == 0:
            print(f"WARNING: No detections found for {model_name}!")
            print(f"Try lowering the confidence threshold (current: {conf_threshold})")

            # Try with an even lower threshold for diagnostic purposes
            if conf_threshold > 0.05:
                very_low_conf = 0.01
                print(f"\nTrying again with very low confidence ({very_low_conf})...")
                results_low_conf = model.predict(source=image_path, conf=very_low_conf, verbose=False)
                low_conf_detections = len(results_low_conf[0].boxes)
                print(f"Detections with {very_low_conf} confidence: {low_conf_detections}")

                if low_conf_detections > 0:
                    print("This suggests the model is detecting objects but with low confidence.")
                else:
                    print("The model is not detecting any objects even at very low confidence.")
        else:
            # 7. Display detailed information about each detection
            print("\nDetection details:")
            for i in range(num_detections):
                class_id = int(boxes.cls[i].item())
                confidence = boxes.conf[i].item()
                coordinates = boxes.xyxy[i].tolist()
                print(f"  Detection #{i+1}:")
                print(f"    Class ID: {class_id}")
                print(f"    Confidence: {confidence:.4f}")
                print(f"    Coordinates (xyxy): {[round(c, 2) for c in coordinates]}")

        # 8. Visualize the results
        image = cv2.imread(image_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        for result in results:
            result_plotted = result.plot()
            plt.figure(figsize=(10, 10))
            plt.imshow(result_plotted)
            plt.title(f"{model_name} Predictions (conf>{conf_threshold})")
            plt.axis('off')
            plt.tight_layout()
            plt.show()

        return num_detections, results

    except Exception as e:
        print(f"ERROR during model debugging: {str(e)}")
        import traceback
        traceback.print_exc()
        return 0, None

# Main debugging script
def main():
    # Test image path
    image_path = "/content/pest24-maryam-1/test/images/0028027_jpg.rf.b6733bf47df381cebb33a3b2cb2a742a.jpg"

    # Debug PlantDoc model
    plantdoc_detections, plantdoc_results = debug_yolo_model(
        "best_plantdoc.pt",
        image_path,
        "PlantDoc",
        conf_threshold=0.6
    )

    # Debug Pest24 model
    pest24_detections, pest24_results = debug_yolo_model(
        "best_pest24.pt",
        image_path,
        "Pest24",
        conf_threshold=0.4
    )

    # Summary
    print("\n" + "="*50)
    print("DEBUGGING SUMMARY")
    print("="*50)
    print(f"PlantDoc detections: {plantdoc_detections}")
    print(f"Pest24 detections: {pest24_detections}")

    if plantdoc_detections > 0 and pest24_detections == 0:
        print("\nPOSSIBLE ISSUES:")
        print("1. The test image may not contain any pests (only plant diseases)")
        print("2. The Pest24 model might not be detecting objects in this image")
        print("3. The confidence threshold might be too high for Pest24 model")
        print("\nSUGGESTIONS:")
        print("- Try using a known pest image that should be detected by Pest24")
        print("- Try running the model with a very low confidence threshold (0.01)")
        print("- Verify the Pest24 model was trained correctly and saved properly")

    elif plantdoc_detections == 0 and pest24_detections > 0:
        print("\nINTERESTING: Only pest detections but no plant disease detections.")

    elif plantdoc_detections == 0 and pest24_detections == 0:
        print("\nBOTH MODELS FAILED TO DETECT OBJECTS!")
        print("Check if the image is valid and contains objects the models should detect.")


if __name__ == "__main__":
    main()
```
